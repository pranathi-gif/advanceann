```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
DATA_PATH=r"/content/Iris.csv"
df=pd.read_csv(DATA_PATH)
df
```

```python
df.isnull().sum()
```

```python
df['Species'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Species']=le.fit_transform(df['Species'])
```

```python
x=df.drop('Species',axis=1)
```

```python
y=df['Species']
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
mc=MinMaxScaler()
x_train=mc.fit_transform(x_train)
x_test=mc.transform(x_test)
```

```python
model=keras.Sequential([layers.Input(shape=(x_train.shape[1],)),
                        layers.Dense(32, activation='relu'),
                        layers.Dense(16,activation='relu'),
                        layers.Dense(3,activation='softmax')])
model.summary()
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history=model.fit(x_train,y_train,epochs=30,batch_size=50,validation_split=0.2,verbose=1)
```

```python
loss,accuracy=model.evaluate(x_test,y_test)
```

```python
plt.plot(history.history['loss'],label='training')
plt.plot(history.history['val_loss'],label='validation')
```

```python
plt.plot(history.history['accuracy'],label='training')
plt.plot(history.history['val_accuracy'],label='validation')
```
